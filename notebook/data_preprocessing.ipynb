```python
import pandas as pd
```

```python
df = pd.read_csv('VoviAI_010.csv', encoding='utf-8')
df.head()
```

```python
df.info()
```

```python
user_df = pd.read_csv('user.csv', encoding='utf-8')
user_df.head()
```

```python
textscript_df = pd.read_csv('text_script.csv', encoding='utf-8')
textscript_df.head()
```

```python
merged_df = pd.merge(df, user_df, left_on='userId', right_on='id', how='inner')
merged_df = pd.merge(merged_df, textscript_df, left_on='textScriptId', right_on='id', how='inner')
merged_df.head()
```

```python
merged_df.info()
```

```python
central = [
    "Thanh Hoa", "Nghe An", "Ha Tinh", "Quang Binh", "Quang Tri", "Thua Thien Hue", "Da Nang",
    "Quang Nam", "Quang Ngai", "Binh Dinh", "Phu Yen", "Khanh Hoa", "Ninh Thuan", "Binh Thuan",
    "Kon Tum", "Gia Lai", "Dak Lak", "Dak Nong", "Lam Dong"
]


north = [
    "Ha Noi", "Ha Giang", "Cao Bang", "Lao Cai", "Bac Kan", "Lang Son", "Tuyen Quang", "Yen Bai",
    "Thai Nguyen", "Phu Tho", "Bac Giang", "Quang Ninh", "Bac Ninh", "Hung Yen", "Hai Duong",
    "Hai Phong", "Vinh Phuc", "Nam Dinh", "Thai Binh", "Ha Nam", "Ninh Binh", "Hoa Binh", "Son La",
    "Dien Bien", "Lai Chau"
]

south = [
    "Ho Chi Minh", "Binh Duong", "Dong Nai", "Ba Ria - Vung Tau", "Tay Ninh", "Binh Phuoc",
    "Long An", "Tien Giang", "Ben Tre", "Tra Vinh", "Vinh Long", "Dong Thap",
    "An Giang", "Kien Giang", "Can Tho", "Hau Giang", "Soc Trang", "Bac Lieu", "Ca Mau"
]
```

```python
new_df = merged_df[['audioLink', 'text', 'userId', 'environment', 'gender', 'accent']].copy()
new_df.head()
```

```python
def get_region(accent):
	if accent in central:
		return 'Central'
	elif accent in north:
		return 'North'
	elif accent in south:
		return 'South'
	else:
		return 'Unknown'

new_df['region'] = new_df['accent'].apply(get_region)
```

```python
new_df.rename(columns={'accent': 'province_name'}, inplace=True)

new_df.head()
```

```python
new_df['gender'] = new_df['gender'].map({'Female': 0, 'Male': 1})

new_df.rename(columns={'userId': 'speakerId'}, inplace=True)

new_df.head(10)
```

```python
new_df.to_csv('VoviAI_010_cleaned.csv', index=False, encoding='utf-8')
```

```python
clean_df = pd.read_csv('VoviAI_010_cleaned.csv', encoding='utf-8')
clean_df.head()
```

```python
import re

def clean_text(text):
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'[“”"‘’•@#$%^&*+=\[\]{}|\\:<>]', '', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip().lower()
    return text

example_text = clean_df['text'].iloc[5]
cleaned_text = clean_text(example_text)

print(f"Original text: {example_text}")
print(f"Cleaned text: {cleaned_text}")
```

```python
clean_df['text'] = clean_df['text'].apply(clean_text)
clean_df.head()
```

```python
clean_df.to_csv('VoviAI_010_cleaned_final.csv', index=False, encoding='utf-8')
```

```python
final_df = pd.read_csv('VoviAI_010_cleaned_final.csv', encoding='utf-8')
final_df.head()
```

```python
from sklearn.model_selection import train_test_split

unique_speakers = clean_df['speakerId'].unique()

train_speakers, temp_speakers = train_test_split(unique_speakers, test_size=0.2, random_state=42)
val_speakers, test_speakers = train_test_split(temp_speakers, test_size=0.5, random_state=42)

train_df = clean_df[clean_df['speakerId'].isin(train_speakers)]
val_df = clean_df[clean_df['speakerId'].isin(val_speakers)]
test_df = clean_df[clean_df['speakerId'].isin(test_speakers)]

print(f"Train set size: {len(train_df)}")
print(f"Validation set size: {len(val_df)}")
print(f"Test set size: {len(test_df)}")
```

```python
train_df.to_csv("train.csv", index=False)
val_df.to_csv("val.csv", index=False)
test_df.to_csv("test.csv", index=False)
```

```python
from datasets import Dataset, DatasetDict

dataset = DatasetDict({
    "train": Dataset.from_pandas(train_df),
    "validation": Dataset.from_pandas(val_df),
    "test": Dataset.from_pandas(test_df)
})
```
